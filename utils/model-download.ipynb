### Download the model from hugginface 'microsoft/beit-base-patch16-224-pt22k-ft22k'

```python
from huggingface_hub import hf_hub_download
import os
import shutil

# Model repo and target directory
repo_id = "microsoft/beit-base-patch16-224-pt22k-ft22k"
target_dir = "/Volumes/KODAK/folder_02/Skin_cancer_Detection/model/Raw_model/beit_base_patch16_model"
os.makedirs(target_dir, exist_ok=True)

# Files to download
files_to_download = [
    "config.json",
    "pytorch_model.bin",
    "preprocessor_config.json",  # equivalent to feature extractor config
    "special_tokens_map.json",   # optional
    "tokenizer_config.json"      # optional
]

# Download each file manually
for filename in files_to_download:
    try:
        print(f"⬇️ Downloading {filename}...")
        file_path = hf_hub_download(repo_id=repo_id, filename=filename)
        shutil.copy(file_path, os.path.join(target_dir, filename))
    except Exception as e:
        print(f"❌ Failed to download {filename}: {e}")

print(f"\n✅ All files downloaded to: {target_dir}")
```

```python
from transformers import BeitForImageClassification, BeitFeatureExtractor

# Path to the saved model directory
model_path = "/Volumes/KODAK/folder_02/Skin_cancer_Detection/model/Raw_model/beit_base_patch16_model"

# Load model and feature extractor
model = BeitForImageClassification.from_pretrained(model_path)
extractor = BeitFeatureExtractor.from_pretrained(model_path)

# Optional: confirm that it loaded successfully
print("✅ Model and extractor loaded.")
print(f"Number of output labels: {model.config.num_labels}")
```
